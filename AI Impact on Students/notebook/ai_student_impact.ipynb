#### 27. How does GPA change across different levels of weekly GenAI usage?

```python
# Calculate the change in GPA for each student
df['GPA_Change'] = (
    df['Post_Semester_GPA'] - df['Pre_Semester_GPA']
)

# Group students by their weekly GenAI usage
df['AI_Usage_Level'] = pd.cut(
    df['Weekly_GenAI_Hours'],
    bins=[-1, 5, 10, 20, 40],
    labels=['0–5 hours', '5–10 hours', '10–20 hours', '20–40 hours']
)

# Calculate average GPA change for each usage level
gpa_change_by_usage = df.groupby(
    'AI_Usage_Level', observed=False
)['GPA_Change'].mean()

print(gpa_change_by_usage)

# Plot the average GPA change
gpa_change_by_usage.plot(kind='bar', figsize=(8, 5))

plt.title('Average GPA Change Across Weekly GenAI Usage Levels')
plt.xlabel('Weekly GenAI Usage')
plt.ylabel('Average GPA Change')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 10. Key Findings and Insights

The exploratory data analysis of 50,000 student records revealed several patterns related to Generative AI usage, academic performance, study habits, skill retention, and student well-being.

1. **Data Quality:** The dataset contains 50,000 records and 16 columns. No missing values, completely duplicated rows, or duplicate Student_ID values were found during the data quality checks.

2. **GenAI Usage:** Weekly GenAI usage varies considerably among students. The average usage was approximately 8.43 hours per week, while the median was 5.8 hours, indicating a right-skewed distribution.

3. **Popular GenAI Applications:** Debugging/Troubleshooting was the most common primary use case (24.6%), followed by Copywriting/Drafting (24.0%). Direct_Answer_Generation was the least common (12.7%).

4. **Academic Performance:** The average pre-semester GPA was approximately 3.15, compared with an average post-semester GPA of 3.35. The average individual GPA change was approximately +0.20 points. GPA increased for 43,759 students, decreased for 6,192, and remained unchanged for 49 students.

5. **AI Dependency and Skill Retention:** The boxplot showed a pattern of lower skill-retention scores at higher perceived AI-dependency levels. This suggests an association worth investigating further, but it does not establish a causal relationship.

6. **Study Habits:** Students spent an average of approximately 11.21 hours per week on traditional studying. The scatter plot comparing traditional study hours with GenAI usage showed a broad spread of observations, suggesting that the relationship is not straightforward.

7. **Student Well-being:** Medium burnout risk was the most common category, accounting for 21,144 students. Students in the high-burnout group also had a higher average exam-anxiety score (4.89) than those in the low-burnout group (3.93).

8. **Prompt-Engineering Skills:** Students classified as advanced in prompt engineering had a slightly higher average post-semester GPA (3.39) than beginner and intermediate students (approximately 3.33 for each group).

9. **Tool Diversity:** Average skill-retention scores increased from 72.59 among students using one tool to 79.81 among students using five tools. This is an observed association and does not prove that using more tools improves retention.

10. **Institutional Policies:** Approximately 50.4% of students were under an institutional policy allowing AI use with citation, 30.0% under actively encouraged use, and 19.6% under a strict ban. Average skill-retention scores differed only modestly across these policy categories.

11. **Paid Subscriptions:** Students with paid GenAI subscriptions reported higher average weekly usage (10.33 hours) than students without paid subscriptions (7.03 hours).

12. **Exam Anxiety:** A total of 7,709 students (15.42%) had an exam-anxiety score of 7 or above, based on the threshold used in this analysis.

Overall, the analysis identified several patterns and associations between AI usage and student-related outcomes. These findings provide a descriptive understanding of the dataset but should not be interpreted as evidence that AI usage directly causes changes in academic performance, skill retention, or well-being.

## 11. Limitations

Although this exploratory data analysis provides useful insights, several limitations should be considered.

1. **Dataset Representativeness:** The analysis is based on the available dataset of 50,000 student records. Without sufficient information about the sampling method and population, the results cannot automatically be generalized to all students.

2. **Observational Analysis:** The analysis identifies patterns and associations but does not establish cause-and-effect relationships. For example, a relationship between AI dependency and skill retention does not prove that AI dependency causes lower retention.

3. **Self-Reported Information:** Variables such as weekly GenAI usage, traditional study hours, perceived AI dependency, and exam anxiety may be subject to reporting or measurement errors if they were self-reported.

4. **Limited Context:** The dataset includes academic, behavioral, and well-being indicators, but other factors—such as teaching methods, course difficulty, prior knowledge, and personal circumstances—may also influence student outcomes.
